```python
# C2.Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy.

#C2
import numpy as np
x = np.array([10, 20, 30, 40, 50])
def standardize(x):
   return (x-np.mean(x))/np.std(x)
print(standardize(x))
```

E1. Per-Column Min–Max Normalization 

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.


```python
# C8.Given 4 RGB images of shape (28, 28, 3), stored as shape (4, 28, 28, 3), compute the average pixel value of each image across height, width and channels. State the axis argument and why.

import numpy as np
x = np.array([[10, 100],[20, 200],[30, 300]])
col_min = np.min(x, axis=0)
col_max = np.max(x, axis=0)
normalized = (x - col_min) / (col_max - col_min)
print(normalized)
```

**Explanation:** axis=0 finds the minimum and maximum of each column separately, and broadcasting applies those values to normalize every column independently.


E2. Three-Level Score Classification

Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as: Fail (< 40), Average (40–70 inclusive), or Excellent (> 70). Hint: You may need to nest or combine np.where() calls


```python
# C8.Given 4 RGB images of shape (28, 28, 3), stored as shape (4, 28, 28, 3), compute the average pixel value of each image across height, width and channels. State the axis argument and why.

import numpy as np
scores = np.array([25, 40, 55, 70, 85, 95])
result = np.where(scores < 40, "Fail",np.where(scores <= 70, "Average", "Excellent"))
print(result)
```

**Explanation:** The nested np.where() first checks for scores below 40. If true, it classifies "Fail", else it checks whether the remaining scores are 70 or below and classifies "Average" if true, otherwise it classifies the rest as "Excellent".
