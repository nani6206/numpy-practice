# NumPy Fundamentals and Operations

Hands-on notes covering NumPy arrays, shapes, indexing, slicing, reshaping, copying vs views, broadcasting, universal functions, statistics, and basic array operations.


```python
# Import NumPy and check the installed version.

import numpy as np
print(f"NumPy version: {np.__version__}")
```

```python
# Create 1D, 2D, and 3D arrays and inspect basic array properties.

import numpy as np
a=np.array([1, 2, 3])
print(a)
b=np.array([[4, 5, 6],[7,8,9]])
print(b)
c=np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(c)
print(f'{a.ndim} {b.ndim} {c.ndim}')
print(f'shape:{c.shape} size:{c.size} dtype:{c.dtype} itemsize:{c.itemsize} nbytes:{c.nbytes}')
```

```python
# Create arrays using zeros(), ones(), full(), arange(), and linspace().

print(f"zeros: {np.zeros((2,3))}")
print(f"ones: {np.ones((2,3))}")
print(f"full: {np.full((2,3),7)}")
print(f"Arrange: {np.arange(0,10,2)}")
print(f"Linspace: {np.linspace(0,1,5)}")
```

```python
# Explore data types, type conversion, and basic indexing.

a=np.array([1,2,3,4,5],dtype=np.int32)
b=np.array([1.5,2.5,3.5],dtype=np.float32)
c=np.array([True,False,True])
print(f"a dtype: {a.dtype} b dtype: {b.dtype} c dtype: {c.dtype}")
x=a.astype(np.float64)
print(x,x.dtype)
print(f"first element of a:{a[0]} using indexing")
```

```python
# Access elements and rows using NumPy indexing and slicing.

matrix=np.array([[1,2,3],
                 [4,5,6]])
print(f"first element of matrix:{matrix[0,0]} using indexing row num and column num")
```

```python
# Reshape a 1D array into 2D and 3D structures.

#for 1-D
print(a[:1])
print(a[-1])
print(a[::-1])
# for 2D
print(matrix[1,:])
```

```python
# Compare views and copies: slicing creates a view, while copy() creates an independent array.

arr=np.arange(1,13)
print(arr)
m=arr.reshape(3,4)
print(m)
cube=arr.reshape(2,2,3)
print(cube)
print(cube.shape)
at=arr.reshape(3,-1)
print(at)
```

```python
# Compare flatten() and ravel(): flatten() returns a copy; ravel() returns a view when possible.

# Compare a view created by slicing with an independent copy.
n = np.arange(1, 6)
view = n[:]
print(n)
view[0] = 10
print(n)

t = np.arange(1, 6)
copy = t.copy()
copy[0] = 10
print(t)
```

```python
# Transpose a matrix and inspect the resulting shape.

# Compare flatten() and ravel() when modifying the resulting 1D array.
print(matrix)

flat = matrix.flatten()
flat[0] = 90
print(matrix)  # Original matrix is unchanged.

rav = matrix.ravel()
rav[0] = 90
print(matrix)  # Original matrix changes when ravel() returns a view.
```

```python
# Demonstrate vectorized arithmetic and scalar operations.

print(matrix)
print(f"Transposed {matrix.T} \nshape:{matrix.T.shape}")
```

```python
# Demonstrate broadcasting between a 2D array and a 1D array.

a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
print(f"vectorised{a+b}")
print(f"vectorised{a-b}")
print(f"vectorised{a*b}")
#scalar operations
print(a)
print(a+5)
print(a-5)
print(a*5)
```

```python
# Create and inspect 2D and 3D arrays.

# adding diff dimension arrays
tuk=np.array([1,2,3])
print(matrix + tuk)
```

```python
# Apply common NumPy universal functions such as sqrt, exp, abs, log, floor, and ceil.

b=np.array([[4, 5, 6],[7,8,9]])
print(b)
c=np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(c)
```

```python
# Calculate common descriptive statistics.

print(f"sqrt:{np.sqrt(a)}")
print(f"exp:{np.exp(a)}")
print(f"abs:{np.abs(np.array([-2,-3,2]))}")
print(f"log:{np.log(np.array([1,np.e,np.e**2]))}")
print(f"floor:{np.floor(np.array([1.2,3.4]))}")
print(f"floor:{np.ceil(np.array([1.2,3.4]))}")
```

```python
# Compare variance and standard deviation for two sets of scores.

scores = np.array([10,20,30,40,50])
print(f"sum:{np.sum(scores)}")
print(f"mean:{np.mean(scores)}")
print(f"median:{np.median(scores)}")
print(f"max:{np.max(scores)}")
print(f"min:{np.min(scores)}")
```

```python
# Aggregate a matrix by columns (axis=0) and rows (axis=1).

classa=[48,49,50,51,52]
classb=[20,35,50,65,80]
print(f"mean:{np.mean(classa)}")
print(f"mean:{np.mean(classb)}")
print(f"variance:{np.var(classa)}")
print(f"var:{np.var(classb)}")
print(f"stddev:{np.std(classa)}")
print(f"stddev:{np.std(classb)}")
```

```python
# Apply comparison operators and basic sorting/selection operations.

print(matrix)
print(np.sum(matrix,axis=0))
print(np.sum(matrix,axis=1))
```
